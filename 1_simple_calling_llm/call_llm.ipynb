```python
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv()
```

```python
model = ChatGoogleGenerativeAI(model="gemini-3.8-flash")

response = model.invoke("Why do parrots talk?")

print(response.text)
```

```python
prompt = "What is the single tagline for a quirky coffee shop. only output the tagline, nothing else"

model = ChatGoogleGenerativeAI(model="gemini-3.8-flash", temperature=0)

response = model.invoke(prompt)

print(response.text)
```

```python
prompt = "What is the single tagline for a quirky coffee shop. only output the tagline, nothing else"

model = ChatGoogleGenerativeAI(model="gemini-3.8-flash", temperature=1)

response = model.invoke(prompt)

print(response.text)
```

```python
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(model="gpt-6-luna")

response = llm.invoke([
    ("system", "You are a helpful assistant that answers in one short line."),
    ("human", "How many moons does Jupiter have?"),
])

print(response.content)
```

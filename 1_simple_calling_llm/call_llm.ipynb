# 🤖 Foundational LLM Calling — Step-by-Step Guide

**What is this module?**
Large Language Models (LLMs) take text input and generate predictive text completions. 
**LangChain** provides a unified, model-agnostic abstraction layer across different AI providers (Google Gemini, OpenAI, Groq, Anthropic), allowing you to swap backends without rewriting application code.

### Architecture Overview

`
┌─────────────────────────┐
│   Environment (.env)    │  Loads API keys securely
└────────────┬────────────┘
             │
             ▼
┌─────────────────────────┐     .invoke()     ┌────────────────────────┐     returns     ┌──────────────────────┐
│      Prompt Input       │ ────────────────► │    Chat Model Engine   │ ──────────────► │   AIMessage Object   │
│ (Raw text or role list) │                   │  (Gemini 2.5 / OpenAI) │                 │  (.content / .text)  │
└─────────────────────────┘                   └───────────┬────────────┘                 └──────────────────────┘
                                                          │
                                                [Temperature Control]
                                                 0.0 = Deterministic
                                                 1.0 = Highly Creative
`

**Key Concepts Covered:**
1. **API Initialization**: Secure key authentication via python-dotenv.
2. **Direct Invocation**: Invoking models with single-string prompts.
3. **Hyperparameter Tuning**: Using 	emperature to control randomness and predictability.
4. **Role-Based Prompting**: Structuring conversations using system instructions and human queries.

## Step 1 — Setup & Environment Configuration

load_dotenv() reads credentials (like GOOGLE_API_KEY and OPENAI_API_KEY) from your local .env file into system environment variables, keeping sensitive secrets out of source code.


```python
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv()
```

---
## Step 2 — Basic LLM Invocation (Google Gemini)

In this step, we initialize ChatGoogleGenerativeAI and send a raw string query.
The model returns an AIMessage response containing generated text and token metadata.


```python
model = ChatGoogleGenerativeAI(model="gemini-3.8-flash")

response = model.invoke("Why do parrots talk?")

print(response.text)
```

---
## Step 3 — Controlling Randomness with Temperature

**What is Temperature?**
Temperature controls the probability distribution over token selection during text generation:

`
Temperature = 0.0 (Argmax / Greedy) ──► Strict, focused, deterministic answers (ideal for code & extraction)
Temperature = 1.0 (High Entropy)   ──► Varied, creative, diverse outputs (ideal for taglines & brainstorming)
`

Below, we test the exact same prompt with 	emperature=0 vs 	emperature=1 to observe the variance.


```python
prompt = "What is the single tagline for a quirky coffee shop. only output the tagline, nothing else"

model = ChatGoogleGenerativeAI(model="gemini-3.8-flash", temperature=0)

response = model.invoke(prompt)

print(response.text)
```

### High Temperature (temperature = 1.0)
Notice how 	emperature=1.0 produces more creative, expressive phrasing compared to the focused 	emperature=0 output above:


```python
prompt = "What is the single tagline for a quirky coffee shop. only output the tagline, nothing else"

model = ChatGoogleGenerativeAI(model="gemini-3.8-flash", temperature=1)

response = model.invoke(prompt)

print(response.text)
```
